```python
# Q1. Write a Python program to print a square star pattern.

n = 5

for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()
```

```python
# Q2. Write a Python program to print a right-angled triangle star pattern.

n = 5

for i in range(1, n + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Q3. Write a Python program to print an inverted right-angled triangle.

n = 5

for i in range(n, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Q4. Write a Python program to print a number triangle.

n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# Q5. Write a Python program to print a triangle with repeated row numbers.

n = 5

for i in range(1, n + 1):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
# Q6. Write a Python program to print an alphabet triangle.

n = 5

for i in range(1, n + 1):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
```

```python
# Q7. Write a Python program to print a pyramid star pattern.

n = 5

for i in range(1, n + 1):
    print(" " * (n - i), end="")
    print("* " * i)
```

```python
# Q8. Write a Python program to print an inverted pyramid star pattern.

n = 5

for i in range(n, 0, -1):
    print(" " * (n - i), end="")
    print("* " * i)
```

```python
# Q9. Write a Python program to print a square number pattern.

n = 5

for i in range(1, n + 1):
    for j in range(1, n + 1):
        print(j, end=" ")
    print()
```

```python
# Q10. Write a Python program to print a decreasing number triangle.

n = 5

for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```
